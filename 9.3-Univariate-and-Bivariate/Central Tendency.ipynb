```python
# Finding the outliers for our dataset
Q1 = dataset[quan].quantile(0.25)
Q3 = dataset[quan].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

# Check
for col in quan:
    less = (dataset[col] < lower[col]).sum()
    great = (dataset[col] > upper[col]).sum()
    print(f"{col}: Lesser={less}, Greater={great}")
```

```python
# Replacing the outliers with median values
import numpy as np
for col in quan:
    median = dataset[col].median()
    dataset[col] = np.where((dataset[col] > upper[col]) | (dataset[col] < lower[col]), median, dataset[col])
```

```python
# After replacing the outliers
for col in quan:
    less = (dataset[col] < lower[col]).sum()
    great = (dataset[col] > upper[col]).sum()
    print(f"{col}: Lesser={less}, Greater={great}")
```

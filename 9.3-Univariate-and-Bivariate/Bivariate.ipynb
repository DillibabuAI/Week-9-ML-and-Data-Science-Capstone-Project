```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
def calc_vif(X):
    #Calculating VIF
    vif = pd.DataFrame()
    vif["variables"] = X.columns
    vif["VIF"] = [variance_inflation_factor(X.values,i) for i in range(X.shape[1])]
    return(vif)
```

```python
from scipy.stats import ttest_ind

# No dropna needed - your data has no missing
# Example 1: Does cbo differ between buggy and clean?
clean = dataset[dataset['defect']==0]['cbo']
buggy = dataset[dataset['defect']==1]['cbo']

result = ttest_ind(clean, buggy, equal_var=False)
print(result)

# Example 2: Loop to test ALL metrics at once
features = ['cbo','wmc','returnQty','loopQty','comparisonsQty','stringLiteralsQty','loc']

for col in features:
    g0 = dataset[dataset['defect']==0][col]
    g1 = dataset[dataset['defect']==1][col]
    t, p = ttest_ind(g0, g1, equal_var=False)[:2]
    print(f"{col}: mean0={g0.mean():.1f}, mean1={g1.mean():.1f}, p={p:.4f}")
```

```python
"""Key insight for your report: In this dataset, ONLY cbo (Coupling Between Objects) shows significant difference between defective 
and non-defective classes. All your Qty metrics are NOT significantly different."""
```

```python
from scipy.stats import ttest_ind

# Correct: Compare defect groups, use Welch's t-test
clean = dataset[dataset['defect']==0]['cbo']
buggy = dataset[dataset['defect']==1]['cbo']

result = ttest_ind(clean, buggy, equal_var=False)
print(f"t={result.statistic:.3f}, p={result.pvalue:.6f}")

# For all features at once
for col in features:
    g0 = dataset[dataset['defect']==0][col]
    g1 = dataset[dataset['defect']==1][col]
    t, p = ttest_ind(g0, g1, equal_var=False)[:2]
    print(f"{col}: p={p:.5f} {'SIGNIFICANT' if p<0.05 else ''}")
```

```python
# For the above case.......
# Reject Null hypothesis if p value < 0.05
# Accept Null hypothesis and reject alternate hypothesis
#Independent Welch's t-test (defect=0 vs defect=1, n=3026 each) showed only 3 of 21 metrics are statistically significant (p<0.05): nosi (p=6.07e-39), 
#dit (p=2.09e-07), and cbo (p=6.81e-07). All Quantity metrics including returnQty (p=0.905), comparisonsQty (p=0.829) and stringLiteralsQty (p=0.941) 
#showed no significant difference between clean and buggy classes.
```

# ANAVO: Analysis of Variance

One way variance of analysis

```python
#Dependent variable is continuous (like cbo, loc, wmc)Independent variable has 3+ categories (like returnQty groups) Groups are independent
#Roughly normal distribution & equal variances

from scipy.stats import f_oneway
import pandas as pd

# No dropna needed - since our data has 0 missing

# Example 1: Does cbo differ across different returnQty levels?
# Create 3 groups for ANOVA

group_low = dataset[dataset['returnQty'] <= 2]['cbo'] # returnQty 0,1,2
group_med = dataset[(dataset['returnQty'] > 2) & (dataset['returnQty'] <= 10)]['cbo'] # 3-10
group_high = dataset[dataset['returnQty'] > 10]['cbo'] # >10

# One-way ANOVA
f_stat, p_value = f_oneway(group_low, group_med, group_high)

print(f"F = {f_stat:.3f}, p = {p_value:.6f}")
if p_value < 0.05:
    print("SIGNIFICANT: At least one group is different")
else:
    print("NOT SIGNIFICANT")
```

```python
# Example 2: Full ANOVA for all metrics at once (clean way for thesis)
features = ['cbo','wmc','loc','comparisonsQty','stringLiteralsQty']

for feature in features:
    g1 = dataset[dataset['returnQty'] <= 2][feature]
    g2 = dataset[(dataset['returnQty'] > 2) & (dataset['returnQty'] <= 10)][feature]
    g3 = dataset[dataset['returnQty'] > 10][feature]

    f, p = f_oneway(g1, g2, g3)
    print(f"{feature}: F={f:.2f}, p={p:.5f} {'*' if p<0.05 else ''}")
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
# For post-hoc
dataset['return_group'] = pd.cut(dataset['returnQty'],
                                 bins=[-1,2,10,600],
                                 labels=['Low (0-2)','Medium (3-10)','High (>10)'])

tukey = pairwise_tukeyhsd(endog=dataset['cbo'],
                          groups=dataset['return_group'],
                          alpha=0.05)

print(tukey)
```

```python
from scipy.stats import f_oneway, tukey_hsd

# Bin returnQty
dataset['return_group'] = pd.cut(dataset['returnQty'], 
                            bins=[-1,2,10,1000], 
                            labels=['Low (0-2)','Medium (3-10)','High (>10)'])

g1 = dataset[dataset['return_group']=='Low (0-2)']['cbo']
g2 = dataset[dataset['return_group']=='Medium (3-10)']['cbo']
g3 = dataset[dataset['return_group']=='High (>10)']['cbo']

f_stat, p_val = f_oneway(g1, g2, g3)
print(f"F={f_stat:.2f}, p={p_val:.4e}")

# Tukey post-hoc
print(tukey_hsd(g1, g2, g3))
```

```python
"""A one-way ANOVA was performed to evaluate whether coupling between objects (cbo) differed across return statement quantity groups. 
ReturnQty was categorized into Low (0-2), Medium (3-10), and High (>10). 
The ANOVA revealed a statistically significant effect, F(2, 6049) = 468.07, p < 0.001, p = 9.84e-190. 
Mean cbo increased with returnQty: Low (M=10.36), Medium (M=17.06), High (M=38.70). 
Post-hoc Tukey HSD tests indicated that all pairwise comparisons were significant (all p < 0.001). 
This suggests that classes with more return statements are significantly more coupled."""
```

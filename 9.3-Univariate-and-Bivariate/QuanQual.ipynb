```python
def quanQual(dataset):
    quan=[]
    qual=[]
    for columnName in dataset.columns:
        #print(columnName)
        if(dataset[columnName].dtype=='O'):
            #print("qual")
            qual.append(columnName)
        else:
            #print("quan")
            quan.append(columnName)
    return quan,qual
```

```python
# Display the quanQual dataset
quanQual(dataset)
```

```python
# Assigning the quan, qual datasets to separate variables
quan,qual=quanQual(dataset)
```

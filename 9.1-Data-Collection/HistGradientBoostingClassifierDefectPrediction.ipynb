```python
# Importing the required library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Loading the dataset
dataset = pd.read_csv("DillibabuSarva_DefectDataset.csv")
dataset
```

```python
# Removing the first column
dataset.drop('SHA', axis = 1, inplace = True)
```

```python
# Displaying the rows and columns count
dataset.shape
```

```python
dataset.columns
```

```python
# Display the independent variable data
independent = dataset[['cbo', 'wmc', 'dit', 'rfc', 'lcom', 'totalMethods', 'totalFields',
       'nosi', 'loc', 'returnQty', 'loopQty', 'comparisonsQty', 'tryCatchQty',
       'parenthesizedExpsQty', 'stringLiteralsQty', 'numbersQty',
       'assignmentsQty', 'mathOperationsQty', 'variablesQty',
       'maxNestedBlocks', 'uniqueWordsQty']]
independent
```

```python
# Display the dependent variable data
dataset[['defect']] = dataset[['defect']].astype('category')
dependent = dataset[['defect']]
print(dependent)
```

```python
# Doing train_test_split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(independent,dependent,test_size = 0.2, random_state = 0)
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.datasets import make_hastie_10_2
classifier = HistGradientBoostingClassifier(max_iter=100).fit(X_train, y_train)
```

```python
# Predicting the values
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
```

```python
print(cm)
```

```python
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, y_pred)
```

```python
#Random Forest classification report
print(clf_report)
```

```python
#importing pickle library for deployment phase
import pickle
```

```python
#here we are assigning the saving model file name with extension to the filename variable
filename = "finalized-model_CapstoneSoftwareDefectPrediction.sav"
#using pickle dump method we are writing the file on disk
pickle.dump(classifier, open(filename, 'wb'))
```

```python
#here we are pedicting the value like we want and assigning to result variable.
loaded_model = pickle.load(open(filename, 'rb'))
result = loaded_model.predict([[14, 7, 1, 11, 3, 3, 1, 0, 32, 5, 0, 2, 0, 0, 2, 0, 4, 0, 5, 1, 38]])
```

```python
#displaying the predicted value for the given 5 values.
result[0]
```

```python
from sklearn.inspection import permutation_importance
# Calculate permutation importance
result = permutation_importance(
    loaded_model, X_train, y_train, n_repeats=10, random_state=42
)

# Create the DataFrame using the mean importances
importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': result.importances_mean
}).sort_values('Importance', ascending=False)

print(importance)
```

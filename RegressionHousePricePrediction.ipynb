```python
# Importing the required library
import pandas as pd
from Univariate import Univariate
```

```python
# Loading the dataset
dataset = pd.read_csv("House_Price_Prediction.csv")
dataset
```

```python
# Displaying the rows and columns count
dataset.shape
```

```python
quan,qual=Univariate.quanQual(dataset)
```

```python
quan
```

```python
qual
```

```python
dataset.info()
```

```python
# Display the quantitative dataset
df = dataset[quan]
df
```

```python
# Display the dataset columns
df.columns
```

```python
# Display the independent variable data
independent = df[['bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot', 'floors',
       'waterfront', 'view', 'condition', 'sqft_above', 'sqft_basement',
       'yr_built', 'yr_renovated', 'price_per_sqft']]
independent
```

```python
# Display the dependent variable data
dependent = df[['price']]
dependent
```

```python
# Doing train_test_split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(independent,dependent,test_size = 0.2, random_state = 0)
```

```python
# Implementing the RandomForestRegressor method
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators = 50, random_state = 0)
regressor.fit(X_train, y_train)
```

```python
# Predicting the values
y_pred = regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test, y_pred)
r_score
```

```python
#importing pickle library for deployment phase
import pickle
```

```python
#here we are assigning the saving model file name with extension to the filename variable
filename = "finalized-model_CapstoneRandomForest.sav"
#using pickle dump method we are writing the file on disk
pickle.dump(regressor, open(filename, 'wb'))
```

```python
#loading the saved model to loaded_variable.
loaded_model = pickle.load(open("finalized-model_CapstoneRandomForest.sav", 'rb'))
```

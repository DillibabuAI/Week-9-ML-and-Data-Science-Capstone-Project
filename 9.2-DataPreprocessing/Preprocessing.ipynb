```python
# Importing the library
import pandas as pd
```

```python
# Loading the dataset
dataset = pd.read_csv("DillibabuSarva_DefectDataset.csv")
```

```python
# Checking the dataset info, which shows count values, datatypes
dataset.info()
```

```python
# Displaying the number of columns and rows
dataset.shape
```

```python
# Checking the missing values to know whether we should perform any data preprocessing techniques or not.
dataset.isna().sum()
```

```python
# Just checking only which are the columns having null values
dataset.columns[dataset.isnull().any()].tolist()
```

```python
# In the above we can see 3 of the columns having null values and preprocessing the data is must here.
dataset['comparisonsQty'] = dataset['comparisonsQty'].fillna(dataset['comparisonsQty'].mean())
dataset['tryCatchQty'] = dataset['tryCatchQty'].fillna(dataset['tryCatchQty'].median())
dataset['stringLiteralsQty'] = dataset['stringLiteralsQty'].fillna(dataset['stringLiteralsQty'].mean())
```

```python
# In the above we used mean, median values to fill the null values.
# Now checking the if still my dataset having any null values.
dataset.columns[dataset.isnull().any()].tolist()
```

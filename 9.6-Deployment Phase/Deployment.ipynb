```python
# Get the library files
import pickle
```

```python
final_features = ['nosi','dit','cbo','rfc','maxNestedBlocks',
            'uniqueWordsQty','assignmentsQty','numbersQty',
            'tryCatchQty','parenthesizedExpsQty']
```

```python
import pandas as pd
dataset = pd.read_csv("DillibabuSarva_DefectDataset.csv")
X_final = dataset[final_features]
y_final = dataset['defect']
```

```python
import warnings
warnings.filterwarnings('ignore')
from sklearn.ensemble import AdaBoostClassifier
final_model = AdaBoostClassifier(n_estimators=200, learning_rate=0.1, random_state=42)
final_model.fit(X_final, y_final)
```

```python
# Finding the outliers for our dataset
Q1 = dataset[final_features].quantile(0.25)
Q3 = dataset[final_features].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
#Check
for col in final_features:
    less = (dataset[col] < lower[col]).sum()
    great = (dataset[col] > upper[col]).sum()
    print(f"{col}: Lesser = {less}, Greater = {great}")
```

```python
deployment_data = {
    'model': final_model,
    'features': final_features,
    'upper': upper, 
    'lower': lower,
    'medians': {col: dataset[col].median() for col in final_features}
}
```

```python
with open("AdaBoost_final_defect_prediction_model.pkl", "wb") as f:
    pickle.dump(deployment_data, f)

print("Ready for deployment!")
```
